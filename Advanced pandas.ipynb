```python
('''Q1 Dataset Exploration

Perform the following exploratory steps:

Display the first 10 rows

Display the dataset structure using info()

Find the number of rows and columns

Identify the data types of each column''')
```

```python
import pandas as pd

# Load the dataset
df = pd.read_csv("Advanced_pandas_sales_dataset.csv")

# 1. Display first 10 rows
print("--- First 10 Rows ---")
print(df.head(10))

# 2. Display the dataset structure using info()
print("\n--- Dataset Structure ---")
df.info()

# 3. Find the number of rows and columns
print(f"\nDataset Shape (Rows, Columns): {df.shape}")

# 4. Identify the data types of each column
print("\n--- Data Types ---")
print(df.dtypes)
```

```python
('''Q2 Sales Filtering Analysis

Filter the dataset to display:

Orders where Sales > 3500

Orders where Region = "East" and Sales > 2500

Explain what insights you observe.''')
```

```python
# 1. Orders where Sales > 3500
sales_above_3500 = df[df['Sales'] > 3500]
print(f"Number of orders with Sales > 3500: {len(sales_above_3500)}")

# 2. Orders where Region = "East" and Sales > 2500
east_high_sales = df[(df['Region'] == 'East') & (df['Sales'] > 2500)]
print(f"Number of orders in East with Sales > 2500: {len(east_high_sales)}")
print(east_high_sales[['Order_ID', 'Customer', 'Product', 'Sales', 'Profit']].head())
```

```python
('''Insights:
Out of 230 total orders, nearly 45% (104 orders) cross the 3,500 sales mark, indicating a robust high-ticket consumer segment.
    Furthermore, the East region alone contributes 48 high-value transactions above 2,500 sales, making it a pivotal area for high-volume high-ticket revenue generation.''')
```

```python
('''Q3 Customer Purchase Analysis

Identify the top 5 customers based on total sales.

Hint: Use GroupBy with aggregation.''')
```

```python
top_customers = df.groupby('Customer')['Sales'].sum().sort_values(ascending=False).head(5)
print("--- Top 5 Customers by Total Sales ---")
print(top_customers)
```

```python
('''Q4 Regional Sales Performance

Calculate the total sales generated by each region.

Then identify:

Which region generated the highest revenue?''')
```

```python
regional_sales = df.groupby('Region')['Sales'].sum().sort_values(ascending=False)
print("--- Regional Sales Performance ---")
print(regional_sales)
print(f"\nHighest Revenue Generating Region: {regional_sales.index[0]} ({regional_sales.max()})")
```

```python
('''Q5 Product Performance Analysis

Calculate the total quantity sold for each product.

Which product appears to be the most popular?''')

```

```python
product_quantity = df.groupby('Product')['Quantity'].sum().sort_values(ascending=False)
print("--- Product Popularity by Total Quantity Sold ---")
print(product_quantity)
print(f"\nMost Popular Product: {product_quantity.index[0]} ({product_quantity.max()} units sold)")
```

```python
('''Q6 Multi-Level Grouping

Group the dataset by: Region and Category

Calculate the total sales for each combination.

What insights can you derive from this analysis?''')
```

```python
multi_group = df.groupby(['Region', 'Category'])['Sales'].sum().unstack()
print("--- Total Sales by Region and Category ---")
print(multi_group)
```

```python
('''Q7 Pivot Table Creation

Create a pivot table showing:

Rows → Region

Columns → Product

Values → Total Sales

Which region sells the most laptops?''')
```

```python
sales_pivot = df.pivot_table(index='Region', columns='Product', values='Sales', aggfunc='sum')
print("--- Pivot Table: Regional Product Sales ---")
print(sales_pivot)
print(f"\nRegion selling the most Laptops: {sales_pivot['Laptop'].idxmax()} ({sales_pivot['Laptop'].max()})")
```

```python
('''Insights:
Electronics is the dominant category across all regions, particularly in the East where it generates an outstanding 160,907 in sales.

Office Supplies lag significantly behind across all zones, marking a potential area where promotional strategies or structural adjustments might be required.''') 
```

```python
('''8 Profitability Analysis

Calculate the average profit for each product category.

Which category is most profitable?''')

```

```python
category_profit = df.groupby('Category')['Profit'].mean().sort_values(ascending=False)
print("--- Average Profit by Category ---")
print(category_profit)
print(f"\nMost Profitable Category (on average): {category_profit.index[0]}")
```

```python
('''Q9 Discount Impact Study

Analyze the relationship between discount and profit.

Tasks:

Calculate the average profit for each discount level

Identify whether higher discounts lead to lower profit''')
```

```python
discount_impact = df.groupby('Discount')['Profit'].mean().sort_values(ascending=False)
print("--- Average Profit by Discount Level ---")
print(discount_impact)
```

```python
Conclusion:
Yes, higher discounts directly cause profits to decline monotonically.
    Offering a 0% discount gives the highest average profitability (751.21), whereas increasing discounts up to 20% severely cripples the bottom line margin down to an average of 90.03.
```

```python
('''Q10 End-to-End Data Wrangling Task

Perform the following workflow:

Filter the dataset where Sales > 3000

Group data by Region and Product

Calculate total sales

Reshape the result into a pivot table

Explain the business insights obtained from this analysis.''')
```

```python
# Step 1: Filter
filtered_df = df[df['Sales'] > 3000]

# Step 2 & 3: Group and Aggregating Sales
grouped_df = filtered_df.groupby(['Region', 'Product'])['Sales'].sum().reset_index()

# Step 4: Pivot reshaped output
final_pivot = grouped_df.pivot(index='Region', columns='Product', values='Sales')
print("--- Reshaped High-Value Pivot Table (Sales > 3000) ---")
print(final_pivot)
```
