```python
#Question 2 : Explain the difference between List and Tuple in Python.#
'''Both List and Tuple are used to store multiple values in Python, but they are different in important ways.

1. Mutability (Main Difference)
List → Mutable (you can change, add, or remove items)
Tuple → Immutable (you cannot change items once created)
 
2. Syntax
List → Uses square brackets [ ]
Tuple → Uses round brackets ( )

3. Performance
Tuple → Faster (because it cannot be changed)
List → Slightly slower

4. Use Case
List → When data can change (e.g., user input, dynamic data)
Tuple → When data should stay fixed (e.g., coordinates, constants)'''

```
