```python
import pandas as pd 
df= pd.read_csv( "employee_productivity_dataset.csv")
```

```python
('''Q1. Handle Missing Values (Basic Cleaning)

Fill missing values using the following:

Age → Median

Salary → Mean

Hours_Worked_Per_Week → Median

Performance_Score → Mean

Display the dataset after handling missing values.''')
```

```python
# Fill missing values
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Salary"] = df["Salary"].fillna(df["Salary"].mean())
df["Hours_Worked_Per_Week"] = df["Hours_Worked_Per_Week"].fillna(
    df["Hours_Worked_Per_Week"].median()
)
df["Performance_Score"] = df["Performance_Score"].fillna(
    df["Performance_Score"].mean()
)

print(df.head())
```

```python
('''Explanation
Median is used for Age and Hours_Worked_Per_Week because it is less affected by outliers.
Mean is used for Salary and Performance_Score.''')
```

```python
('''Q2. Label Encoding

Convert these columns into numeric values using Label Encoding:

Gender

Department

Show the updated columns.''')
```

```python
from sklearn.preprocessing import LabelEncoder

le_gender = LabelEncoder()
le_department = LabelEncoder()

df["Gender"] = le_gender.fit_transform(df["Gender"])
df["Department"] = le_department.fit_transform(df["Department"])

print(df[["Gender", "Department"]].head())
```

```python
('''Q3. One-Hot Encoding

Apply One-Hot Encoding on:

Work_Mode

Location

Display the dataset and check how many new columns are created''')
```

```python
df = pd.get_dummies(
    df,
    columns=["Work_Mode", "Location"]
)

print(df.head())

new_columns = [col for col in df.columns
               if col.startswith("Work_Mode_")
               or col.startswith("Location_")]

print("New Columns Created:", len(new_columns))
print(new_columns)
```

```python
('''Q4. Normalization (Min-Max Scaling)

Normalize the following columns:

Salary

Hours_Worked_Per_Week

Use MinMaxScaler and display results''')
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[["Salary", "Hours_Worked_Per_Week"]] = scaler.fit_transform(
    df[["Salary", "Hours_Worked_Per_Week"]]
)

print(df[["Salary", "Hours_Worked_Per_Week"]].head())
```

```python
Explanation

Min-Max Scaling converts values into the range:

x
′
= Xmax −Xmin/ X−Xmin
Range becomes 0 to 1.
```

```python
('''Q5. Standardization (Scaling)

Apply StandardScaler on:

Age
Projects_Completed

Display the transformed values''') 
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[["Age", "Projects_Completed"]] = scaler.fit_transform(
    df[["Age", "Projects_Completed"]]
)

print(df[["Age", "Projects_Completed"]].head())
```

```python
('''Q6. Compare Scaling Methods

Apply both:

MinMaxScaler

StandardScaler

On the Salary column.

Show both results side by side. ''') 
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

comparison = pd.DataFrame()

comparison["Salary"] = df["Salary"]

comparison["MinMaxScaler"] = MinMaxScaler().fit_transform(
    comparison[["Salary"]]
)

comparison["StandardScaler"] = StandardScaler().fit_transform(
    comparison[["Salary"]]
)

print(comparison.head())
```

```python
('''Q7. Build Preprocessing Pipeline

Create a pipeline that:

Applies encoding to categorical columns

Applies scaling to numerical columns''') 
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

categorical_features = [
    "Gender",
    "Department",
    "Work_Mode",
    "Location"
]

numerical_features = [
    "Age",
    "Salary",
    "Hours_Worked_Per_Week",
    "Projects_Completed",
    "Performance_Score"
]

preprocessor = ColumnTransformer(
    transformers=[
        (
            "cat",
            OneHotEncoder(handle_unknown="ignore"),
            categorical_features
        ),
        (
            "num",
            StandardScaler(),
            numerical_features
        )
    ]
)

pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor)
    ]
)

print(pipeline)
```

```python
('''Q8. Apply Pipeline

Apply the pipeline on the dataset.

Display:

Transformed dataset

Shape of final dataset''') 
```

```python
df = pd.read_csv("employee_productivity_dataset.csv")
X = df[
    categorical_features +
    numerical_features
]

X_transformed = pipeline.fit_transform(X)

print("Transformed Dataset:")
print(X_transformed)

print("Shape of Final Dataset:")
print(X_transformed.shape)
```
