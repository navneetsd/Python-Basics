```python
# 1. Display the first 5 rows
print("First 5 rows:")
display(df.head())

# 2. Display the last 5 rows
print("\nLast 5 rows:")
display(df.tail())

# 3. Show the structure of the dataset using info()
print("\nDataset Structure:")
df.info()

# 4. Display the shape of the dataset
print("\nDataset Shape:")
print(df.shape)
```

```python
('''Q4. Indexing using .loc

Using .loc, perform the following:

Display the row with index 10.

Display the Sales value of the row with index 20''')
```

```python
# 1. Display the row with index 10
print("Row at index 10:")
print(df.loc[10])

# 2. Display the Sales value of the row with index 20
print("\nSales value at index 20:")
print(df.loc[20, 'Sales'])
```

```python
('''Q5. Indexing using .iloc

Using .iloc, display:

The first 6 rows and first 4 columns.

The Sales value from the 3rd row.''')
```

```python
# 1. First 6 rows and first 4 columns
print("First 6 rows and first 4 columns:")
display(df.iloc[:6, :4])

# 2. Sales value from the 3rd row (index 2)
# The 'Sales' column is at column position index 4
print("\nSales value from the 3rd row:")
print(df.iloc[2, 4])
```

```python
('''Q6. Row Slicing

Display:

The first 10 rows of the dataset.

Rows 15 to 25.''')
```

```python
# 1. The first 10 rows
print("First 10 rows:")
display(df.iloc[:10])

# 2. Rows 15 to 25 (inclusive of index 25)
print("\nRows 15 to 25:")
display(df.loc[15:25])
```

```python
('''Q7. Filtering Data

Filter the dataset to display:

Rows where Region = "East".

Rows where Sales > 1500.''') 
```

```python
# 1. Filter by Region = "East"
print("Rows where Region is East:")
display(df[df['Region'] == 'East'].head())

# 2. Filter by Sales > 1500
print("\nRows where Sales > 1500:")
display(df[df['Sales'] > 1500].head())
```

```python
('''Q8. Multiple Condition Filtering

Display rows where:

Region is West

Sales is greater than 1200''') 
```

```python
# Multiple conditional filtering using &
filtered_west_sales = df[(df['Region'] == 'West') & (df['Sales'] > 1200)]
filtered_west_sales.head()
```

```python
('''Q9. Descriptive Statistics

Generate summary statistics using:

describe()

From the result identify:

Mean sales value

Maximum quantity sold

Minimum sales value ''')
```

```python
# Summary statistics
summary_stats = df.describe()
print("Summary Statistics:")
display(summary_stats)

# Identifying requested metrics
print(f"Mean sales value: {summary_stats.loc['mean', 'Sales']:.2f}")
print(f"Maximum quantity sold: {summary_stats.loc['max', 'Quantity']}")
print(f"Minimum sales value: {summary_stats.loc['min', 'Sales']}")
```

```python
('''Q10. Aggregation Analysis

Using the Sales column, calculate:

Total sales

Average sales

Maximum sales

Minimum sales ''')
```

```python
# Performing individual aggregations on the Sales column
total_sales = df['Sales'].sum()
avg_sales = df['Sales'].mean()
max_sales = df['Sales'].max()
min_sales = df['Sales'].min()

print(f"Total sales: {total_sales}")
print(f"Average sales: {avg_sales:.2f}")
print(f"Maximum sales: {max_sales}")
print(f"Minimum sales: {min_sales}")
```
